```python
# Since the duplicated data is too low ie. about0.7% so we can drop it 
sales.drop_duplicates(inplace=True)
sales.shape
```

```python
# Duplicated data removed sucessfully
(sales.duplicated().sum()/len(sales))*100
```

```python
# we can see that in col status and unnamed 1 all the values are NaN values only so we can remove these columns
# And about the NaN values of Amount col again it's very less amount so we can remove those rows
sales.isna().sum()
```

```python
sales.drop(['Status','unnamed1'],axis = 1,inplace=True)
```

```python
sales.dropna(inplace=True)
```

```python
sales.shape
```

```python
# Now the columns Gender and Marital_statue contain data like M,F and 0,1 so we need to rewrite those values
# First gender col
# I am using map(), there are other ways too
gen_dict = {'M':'Male','F':'Female'}
sales['Gender'] = sales['Gender'].map(gen_dict)
sales['Gender']
```

```python
sales['Gender'].unique()
```

```python
# Now Marital_Status column
# I am using replace this for this
# But the data is not in string so first i will convert it into str
sales['Marital_Status'] = sales['Marital_Status'].astype(str)
sales['Marital_Status'] = sales['Marital_Status'].str.replace('0','Single')
sales['Marital_Status'] = sales['Marital_Status'].str.replace('1','Married')
```

```python
sales['Marital_Status'].unique()
```

```python
# No of customers wrt gender
sales['Gender'].value_counts()
# the no of female customers is more than male customers more than twice
```

```python
# Total no of orders in each genders
b = sales.groupby(['Gender','Orders'])
b['User_ID'].count()
# The values are almost same in terms of no of orders
```

```python
a = sales.groupby('Gender')
a['Amount'].sum()
# We can see that the total amount spend by female customers is more than twice the total amount spend by male costumers
```

```python
c = sales.groupby('Age Group')
c['Amount'].mean()
# Means all the average amount spent by all the age group is almost the same
```

```python
d = sales.groupby('State')
top_states = d['Amount'].sum().sort_values(ascending=False)
top_5_states = top_states[0:5]
top_states

# Uttar pradesh, maharashtra , Karnataka, Delhi and Madhya Pradesh  are the top 5 staes with highest Amount spend according to this
```

```python
# Creating a function for bar label
def label(a):
    for i in a.containers:
        a.bar_label(i)
```

```python
# no of customer by gender
a = sns.countplot(x='Gender',data=sales)
# We can use histplot too
#sns.histplot(x='Gender',data=sales)
label(a)
plt.show()
```

```python
# Total sales by age group
b = sns.barplot(x='Age Group',y='Amount',data=sales,errorbar=None, estimator= sum)
label(b)
```

```python
#the top 5 states by sales.
plt.figure(figsize=(15,5))
c= sns.barplot(x='State',y='Amount',data=sales,errorbar=None,estimator=sum)
plt.xticks(rotation = 45)
plt.grid()
label(c)
plt.show()
# i used estimator because by default it calculates mean
# other wise we could had used group by and doing sum in it and thendata frame
```

```python
# Only top 5
top_5_states = pd.DataFrame(top_5_states)
top_5_states.reset_index(inplace=True)
plt.figure(figsize=(15,5))
c= sns.barplot(x='State',y='Amount',data=top_5_states,errorbar=None)
plt.xticks(rotation = 45)
plt.grid()
label(c)
plt.show()
```

```python
# sales by product category
plt.figure(figsize=(10,5))
gr_product_cat = sales.groupby('Product_Category')['Amount'].sum()
gr_product_cat = pd.DataFrame(gr_product_cat)
gr_product_cat.reset_index(inplace=True)
d = sns.barplot(x='Product_Category',y='Amount',data=gr_product_cat,errorbar=None)
plt.xticks(rotation = 70)
label(d)
plt.show()
```
